#### There is a big chunk (~20%) of missing data for Episode length and guest popularity. Hence we cannot simply drop it
plot these two data for better understanding

### First plot of episode length

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Boxplot
data.boxplot(column='Episode_Length_minutes', ax=axes[0])
axes[0].set_ylabel('Podcast length in mins')
axes[0].set_title('Box Plot')

# Histogram + KDE
sns.histplot(data['Episode_Length_minutes'].dropna(), kde=True, bins=30, ax=axes[1])
axes[1].set_title('Distribution of Episode Length')
axes[1].set_xlabel('Mins')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Then plot of guest popularity

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Boxplot
data.boxplot(column='Guest_Popularity_percentage', ax=axes[0])
axes[0].set_ylabel('Popularity percent')
axes[0].set_title('Box Plot')

# Histogram + KDE
sns.histplot(data['Guest_Popularity_percentage'].dropna(), kde=True, bins=30, ax=axes[1])
axes[1].set_title('Distribution of Popularity')
axes[1].set_xlabel('Percent')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Lets see the correlation between the columns

```python

plt.figure(figsize=(10, 8))  # adjust size as needed
sns.heatmap(data.drop(columns='id').corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

The only actual corelation seen is between listening time and the episode length
which means epiode length is an important column to fill, popularity is not of much use

We must also encode other non numeric values and check for better understanding

```python
data.dtypes
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df_encoded = data.copy()
label_encoders = {}

# Loop through all non-numeric columns
for col in df_encoded.select_dtypes(exclude='number').columns:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))
    label_encoders[col] = le  # Optional: store encoder for decoding later
```
